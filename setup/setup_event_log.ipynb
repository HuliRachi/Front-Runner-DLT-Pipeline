```python
# COMMAND ---------- NOTEBOOK

dbutils.widgets.text("catalog", "dev", "Target catalog")
dbutils.widgets.text("ingestion_pipeline_id", "2812f55c-dbf8-4baa-a26c-5a3f14335ed9", "frontrunner_ingestion_pipeline ID")
dbutils.widgets.text("transformation_pipeline_id", "64a3d966-ca68-4af7-9e85-cd0adc5a12b2", "frontrunner_transformation_pipeline ID")

catalog = dbutils.widgets.get("catalog")
ingestion_pipeline_id = dbutils.widgets.get("ingestion_pipeline_id")
transformation_pipeline_id = dbutils.widgets.get("transformation_pipeline_id")

if not ingestion_pipeline_id or not transformation_pipeline_id:
    raise ValueError(
        "Both pipeline IDs are required. Find them in each pipeline's Settings panel "
        "in the Lakeflow Pipelines Editor — this notebook won't guess them."
    )


spark.sql(f"""
CREATE OR REPLACE VIEW {catalog}.frontrunner.ingestion_event_log_raw AS
SELECT * FROM event_log("{ingestion_pipeline_id}")
""")

spark.sql(f"""
CREATE OR REPLACE VIEW {catalog}.frontrunner.transformation_event_log_raw AS
SELECT * FROM event_log("{transformation_pipeline_id}")
""")

print(f"Created {catalog}.frontrunner.ingestion_event_log_raw")
print(f"Created {catalog}.frontrunner.transformation_event_log_raw")
```
